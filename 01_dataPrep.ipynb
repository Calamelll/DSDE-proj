# Data Preparation
ทำ data cleansing สำหรับข้อมูล bangkok_traffy.csv เพื่อสร้างโมเดลทำนายการแก้ไขปัญหาเวลาในการทำ

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('bangkok_traffy.csv')
```

```python
df_clean = df
```

```python
# ค้นหาคำว่า "เขค/อำเภอ" ตามด้วยช่องว่าง(มีหรือไม่มีก็ได้) แล้วดึงข้อความข้างหลังออกมา
# (?<=...) คือ Lookbehind เพื่อเช็คคำนำหน้าแต่ไม่เก็บคำนำหน้านั้นมา
pattern_district = r'(?:เขต|อำเภอ|อ\.)\s*([ก-๙a-zA-Z0-9\-\.]+)'

# สร้างตัวแปรชั่วคราวที่ดึงค่าเขตมาจาก Address
extracted_dist = df_clean['address'].str.extract(pattern_district, expand=False)

# เติมลงในช่องที่เป็น NaN เท่านั้น
df_clean['district'] = df_clean['district'].fillna(extracted_dist)

# บางครั้งอาจมีช่องว่างติดมา ให้ตัดออก
df_clean['district'] = df_clean['district'].str.strip()
df_clean.dropna(subset=['district'])

```

เติมข้อมูล District ที่เป็นค่าว่าง (NaN) โดยใช้ Regex ดึงชื่อเขตหรืออำเภอมาจากคอลัมน์ Address แล้วตัดช่องว่างส่วนเกินออก

```python
#Drop columns ที่ไม่จำเป็นออก
cols_to_drop = [
    "photo",
    "photo_after",   
    "address",
    "subdistrict",
    "province",
    "state",          
]
df_clean = df_clean.drop(columns=cols_to_drop, errors="ignore")
df_clean.columns = df_clean.columns.str.strip()
df_clean["star"] = df_clean["star"].fillna(0)
df_clean["count_reopen"] = df_clean["count_reopen"].fillna(0)
df_clean["organization"] = df_clean["organization"].fillna("Unknown")
```

ลดมิติข้อมูล (Drop Unnecessary Columns): ตัดคอลัมน์ที่ไม่ส่งผลดีต่อโมเดลออก
    
    - ลด Noise: ตัด ticket_id (ไม่มีความหมายเชิงพฤติกรรม) และ address/subdistrict (มีความหลากหลายสูง/High Cardinality) เพื่อป้องกันโมเดลจดจำข้อมูลจำเพาะเกินไป (Overfitting)  
    - ป้องกัน Data Leakage: ตัด photo_after เนื่องจากเป็นข้อมูลที่เกิดขึ้น หลัง การทำนาย
    - ขอบเขตงาน: ตัด photo เนื่องจากโปรเจกต์นี้ไม่ครอบคลุมการประมวลผลภาพ (Image Processing)
จัดรูปแบบชื่อคอลัมน์ (Standardize Headers): ตัดช่องว่างส่วนเกินเพื่อความถูกต้องในการเรียกใช้งาน
จัดการค่าสูญหาย (Imputation): แทนที่ star ที่เป็นค่าว่างด้วย 0 (อนุมานว่าไม่มีการให้คะแนน) แทนที่ count_reopen ที่เป็นค่าว่างด้วย 0 และแทนที่ organization เป้น "Unknown"

```python
#Prep For interval
df_clean['timestamp'] = pd.to_datetime(df_clean['timestamp'], errors='coerce', utc=True)
df_clean['timestamp'].dtype
df_clean = df_clean.dropna(subset=["timestamp"])
df_clean['day'] = df_clean['timestamp'].dt.day_name().str[:3]   # Mon, Tue, Wed...
df_clean['hour'] = df_clean['timestamp'].dt.hour                # 0–23
df_clean["month"] = df_clean["timestamp"].dt.month
df_clean["year"] = df_clean["timestamp"].dt.year
```

เตรียมข้อมูลเวลา (Temporal Feature Engineering) โดยคัดกรองข้อมูลที่เวลาไม่ถูกต้องออก และแตกย่อย Timestamp เป็นฟีเจอร์ย่อย (วัน, ชั่วโมง, เดือน, ปี) เพื่อให้โมเดลจับ Pattern ของช่วงเวลาได้ละเอียดขึ้น

```python
coords_split = df_clean["coords"].str.split(",", expand=True)
df_clean["lat"] = pd.to_numeric(coords_split[0], errors="coerce")
df_clean["lon"] = pd.to_numeric(coords_split[1], errors="coerce")
df_clean = df_clean.drop(columns=["coords"])
```

แยกค่า Latitude , Longitude ใน column 'coords' มาใส่ใน column 'lat' และ 'lon' ที่สร้างใหม่ตามลำดับ

```python
# นำมาสร้าง target 
df_clean["last_activity"] = pd.to_datetime(df_clean["last_activity"], errors="coerce", utc=True)
df_clean["resolution_time_hours"] = (df_clean["last_activity"] - df_clean["timestamp"]).dt.total_seconds()/3600

# แล้วค่อย drop ทิ้ง
df_clean = df_clean.drop(columns=["last_activity"], errors="ignore")

# สร้าง column isLate ขึ้นมาโดยกำหนดให้เกิน 7 วัน แปลว่าช้า

# ตั้ง threshold (หน่วย: ชั่วโมง)
late_threshold = 168  # 7 วัน
# สร้างคอลัมน์ isLate (1 = ช้า, 0 = ทันเวลา)
df_clean["isLate"] = (df_clean["resolution_time_hours"] > late_threshold).astype(int)

df_clean["isLate"].value_counts(normalize=True)
```

เนื่องจากเราจะทำฌมเดลเพื่อทำนายว่าการแก้ปัญหาในแต่ละ ticket_id จะช้าหรือทำทันเวลา เราได้สร้าง column 'resolution_time_hours' 
เพื่อมาทำ column 'isLate' โดย isLate จะเป็น 1 เมื่อ resolution_time_hours มีค่ามากกว่า 168 ชั่วโมง (7วัน)

```python
urgency_map = {
    'ความปลอดภัย': 5, 'น้ำท่วม': 5, 'สะพาน': 5, 'สายไฟ': 5,
    'การเดินทาง': 4, 'ถนน': 4, 'ท่อระบายน้ำ': 4, 'แสงสว่าง': 4,
    'PM2.5': 3, 'เสียงรบกวน': 3, 'คนจรจัด': 3, 'กีดขวาง': 3, 'ทางเท้า': 3,
    'ความสะอาด': 2,'สัตว์จรจัด': 2, 'คลอง': 2, 'ห้องน้ำ': 2, 'ต้นไม้': 2, 'จราจร': 2,
    'ป้าย': 1, 'ป้ายจราจร': 1 ,'ร้องเรียน':1, 'แจ้งปัญหา':1, 'เสนอแนะ':1, 'สอบถาม':1
}


def calculate_urgency(type_string):
    if pd.isna(type_string):
        return 1 # ถ้าเป็นค่าว่าง (NaN) ให้ค่าต่ำสุด (หรือจะแก้เป็น 2.5 ด้วยก็ได้ครับ)
    
    # 1. Clean string
    cleaned_string = type_string.replace('{', '').replace('}', '')
    
    # 2. Split
    tags = cleaned_string.split(',')
    
    # 3. Map & Max (จุดที่แก้ไข)
    # เปลี่ยนเลข 1 ด้านหลังเป็น 2.5
    scores = [urgency_map.get(tag.strip(), 2.5) for tag in tags if tag.strip() != '']
    
    # กรณี input เป็น {} ว่างๆ ไม่มีข้อความข้างใน
    if not scores:
        return 1 # หรือจะเปลี่ยนเป็น 2.5 ถ้าต้องการให้เคสว่างๆ มีค่ากลางๆ
        
    return max(scores)

df_clean['urgency_score'] = df_clean['type'].apply(calculate_urgency)
```

เนื่องใน column 'type' เก็บข้อมูลประเภทการเร้องเรียงต่างเป็น set เราจึง map ค่าลำดับความสำคัญ กับเหตุการณ์ที่ร้องเรียงเรียนมาสร้างเป็น column 'urgency_score' (เป็นตัวบ่งบอกว่าเหตุการณืที่ร้องเรียนมาสำคัญไหม)

```python
import os
out_dir = 'data'
os.makedirs(out_dir, exist_ok=True)
out_path = os.path.join(out_dir, 'bangkok_traffy_clean.csv')
df_clean.to_csv(out_path, index=False, encoding='utf-8-sig')
print('Saved cleaned traffy to', out_path)
```
